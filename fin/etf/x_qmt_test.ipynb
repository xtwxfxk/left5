```python
import time
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from xtquant import xtdata

import akshare as ak
```

```python
['159864.SZ', '516880.SH']
# ['159547.SZ', '515450.SH']
## ['159783.SZ', '588390.SH']
## ['159782.SZ', '588350.SH'] *
# ['159861.SZ', '516070.SH']
## ['159779.SZ', '562950.SH']
## ['159560.SZ', '516920.SH']
## ['159761.SZ', '516710.SH']
## ['159828.SZ', '512170.SH']
# ['159838.SZ', '512120.SH']
## ['159602.SZ', '563000.SH'] **
codes = ['159864.SZ', '516880.SH']
```

```python
for code in codes:
    xtdata.download_history_data(code, period='1d', start_time='20230101')
```

```python
history_data = xtdata.get_market_data_ex([], codes, period='1d', count=-1)
```

```python
df1 = pd.DataFrame(history_data[codes[0]])
df2 = pd.DataFrame(history_data[codes[1]])
```

```python
df1['datetime'] = pd.to_datetime(df1['time'] / 1000, unit='s')
df2['datetime'] = pd.to_datetime(df2['time'] / 1000, unit='s')
```

```python
plt.close()
```

```python
plt.plot(df1.close, label=codes[0])
plt.plot(df2.close, label=codes[1])
plt.plot(df2.close - df1.close, label='xxx')
plt.legend()
plt.show()
```

```python
plt.close()
```

```python
fig, ax1 = plt.subplots()

ax1.plot(df1.close, label=codes[0], color='r')
ax1.plot(df2.close, label=codes[1], color='g')
# ax1.xticks(ticks=ticks, labels=tick_labels)

ax2 = ax1.twinx()
ax2.plot(df2.close - df1.close, color='b', label='sub')


lines, labels = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax2.legend(lines + lines2, labels + labels2, loc=0)

fig.tight_layout()
plt.show()
```
