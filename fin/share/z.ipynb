```python
stock_info_sz_name_code_df = ak.stock_info_sz_name_code(symbol="A股列表")
```

```python
stock_info_sz_name_code_df = stock_info_sz_name_code_df[(stock_info_sz_name_code_df['板块'] == '主板') | (stock_info_sz_name_code_df['板块'] == '创业板')]
stock_info_sz_name_code_df = stock_info_sz_name_code_df[~stock_info_sz_name_code_df['A股简称'].str.startswith('ST') & ~stock_info_sz_name_code_df['A股简称'].str.startswith('*ST')]
```

```python
stock_info_sz_name_code_df
```

```python
symbol1 = stock_info_sh_name_code_df[['证券代码', '证券简称', '上市日期']].rename(columns={'证券代码': 'symbol', '证券简称': 'name', '上市日期': 'date'})
symbol1['market'] = 'sh'
```

```python
symbol2 = stock_info_sz_name_code_df[['A股代码', 'A股简称', 'A股上市日期']].rename(columns={'A股代码': 'symbol', 'A股简称': 'name', 'A股上市日期': 'date'})
symbol2['market'] = 'sz'
```

```python
symbols = pd.concat((symbol1, symbol2), axis=0)
```

```python
symbols
```

```python
stock_zh_a_spot_em_df = stock_zh_a_spot_em_df[stock_zh_a_spot_em_df['代码'].isin(symbols.symbol)]
```

```python
stock_zcfz_em_df1 = ak.stock_zcfz_em(date='20230331')
stock_zcfz_em_df2 = ak.stock_zcfz_em(date='20230630')
stock_zcfz_em_df3 = ak.stock_zcfz_em(date='20230930')
stock_zcfz_em_df4 = ak.stock_zcfz_em(date='20231231')
```

```python
stock_zcfz_em_year_df = pd.concat((stock_zcfz_em_df1, stock_zcfz_em_df2, stock_zcfz_em_df3, stock_zcfz_em_df4), axis=0)
```

```python
stock_zcfz_em_df = stock_zcfz_em_year_df[stock_zcfz_em_year_df.groupby('股票代码')['公告日期'].transform('max') == stock_zcfz_em_year_df['公告日期']]
```

```python
debt_df = stock_zcfz_em_df[stock_zcfz_em_df['资产负债率'] < 30]
```

```python
debt_df # [debt_df['股票代码'] == '300192']
```
