```python
import time
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
names = ['time', 'open', 'high', 'low', 'close', 'all_grad_mean', 'all_grad_std', 
         'pike_grad1', 'pike_grad2', 'pike_grad3', 'pike_ma_grad1', 'pike_ma_grad2', 'pike_ma_grad3',
         'sill_grad1', 'sill_grad2', 'sill_grad3', 'sill_ma_grad1', 'sill_ma_grad2', 'sill_ma_grad3',]
```

```python
df = pd.read_csv('D:/grads.csv', header=None, names=names)
```

```python
df['datetime'] = pd.to_datetime(df['time'], format='%Y.%m.%d %H:%M')
```

```python
df = df.set_index('datetime')
```

```python
df
```

```python
fig, ax1 = plt.subplots()

ax1.plot(df.close, label='close', color='r')
# ax1.xticks(ticks=ticks, labels=tick_labels)

ax2 = ax1.twinx()
ax2.plot(df.all_grad_mean, color='g', label='all_grad_mean')
ax2.plot(df.all_grad_std, color='b', label='all_grad_std')
# ax2.plot(df.gh4, color='y', label='gh4')
# plt.xticks(ticks=ticks, labels=tick_labels)
# plt.xticks(ticks=df.datetime)

lines, labels = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax2.legend(lines + lines2, labels + labels2, loc=0)

fig.tight_layout()
plt.show()
```
