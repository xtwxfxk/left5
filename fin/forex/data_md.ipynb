```python
import os, datetime
import pandas as pd
```

```python
root = 'D:/code/mt5/md'
```

```python
datas_eurusd = []
datas_audusd = []

for index, line in enumerate(open(os.path.join(root, 'md.txt'), 'r', encoding='utf-8'), 1):
    a = [l.strip() for l in line.strip().split() if l.strip()]

    symbol = a[5]
    
    t_open = '%s %s' % (a[1], a[2])
    t_close = '%s %s' % (a[9], a[10])

    price_open = a[6]
    volume_open = a[4]
    
    direction_open = a[3]
    direction_close = 'sell' if 'buy' == direction_open else 'buy'

    ticket_open = str(index)
    # print(symbol)
    if 'eurusd' == symbol:
        datas_eurusd.append(','.join([t_open, direction_open, price_open, volume_open, ticket_open]))
        datas_eurusd.append(','.join([t_close, direction_close, price_open, volume_open, ticket_open]))
        
    else:

        datas_audusd.append(','.join([t_open, direction_open, price_open, volume_open, ticket_open]))
        datas_audusd.append(','.join([t_close, direction_close, price_open, volume_open, ticket_open]))

datas_eurusd.sort()
datas_audusd.sort()

with open(os.path.join(root, 'md_eurusd.txt'), 'w', encoding='utf-8') as f:
    f.write('\n'.join(datas_eurusd))

with open(os.path.join(root, 'md_audusd.txt'), 'w', encoding='utf-8') as f:
    f.write('\n'.join(datas_audusd))
```
