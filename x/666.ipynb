```python
selector = Selector(response.text)
```

```python
eles = selector.xpath('//li/dl/div[8]')
```

```python
len(eles)
```

```python
eles[0].xpath('.//span/text()').extract()[0]
```

```python
eles[0].xpath('.//b/text()').extract()[0].encode("latin1").decode("gbk") # .encode('utf-8').decode("gb2312")
```

```python
ages = []
zoos = []
for ele in eles[:-40]:
    age = int(ele.xpath('.//span/text()').extract()[0])
    ages.append(age)

    z = ele.xpath('.//b/text()').extract()[0].encode("latin1").decode("gbk")
    zoos.append(z)
```

```python
# ages
# zoos
```

```python
# sns.displot(ages) # , kind='kde')

plt.figure(figsize=(16, 10))
sns.countplot(x=ages, palette='pastel')

# 添加标签和标题
plt.xlabel('数字')
plt.ylabel('出现次数')
plt.title('数字出现次数统计')
```

```python
plt.figure(figsize=(16, 10))
sns.countplot(x=zoos, palette='pastel')

# 添加标签和标题
plt.xlabel('数字')
plt.ylabel('出现次数')
plt.title('呵呵')
```
